```python
from pyspark.sql import Row
import random

data = [Row(id=i, value=random.randint(1, 100)) for i in range(10)]
df = spark.createDataFrame(data)
display(df)
```

```python
from pyspark.sql.functions import col

# Filter rows where value > 50 and add a new column 'value_squared'
df_transformed = df.filter(col("value") > 50).withColumn("value_squared", col("value") ** 2)
display(df_transformed)
```

```python
from pyspark.sql.functions import avg, min, max, count

# Aggregate the dataframe: count, min, max, and average of 'value'
df_agg = df.agg(
    count("value").alias("count"),
    min("value").alias("min_value"),
    max("value").alias("max_value"),
    avg("value").alias("avg_value")
)
display(df_agg)
```
